```
# Actividad 2.2 - Depuración de Registros de Pedidos
import pandas as pd
import os

df = pd.read_csv("ventas_erroneas_completo.csv", header=None, dtype=str)
cols = ["id_pedido","producto","cantidad","precio_unitario","cliente","fecha_hora","ciudad"]
df.columns = cols
df.head()
```

```
# Inspección
print(df.shape); print(df.duplicated().sum()); df.info()
```

```
# Limpiar valores sucios
m={"nan":None,"NaN":None,"null":None,"NULL":None,"error":None,"cincuenta":"50"}
for k,v in m.items(): df=df.replace(k,v,regex=False)
df["cantidad"]=df["cantidad"].astype(str).str.strip()
df["precio_unitario"]=df["precio_unitario"].astype(str).str.strip()
df.head()
```

```
# Normalizar fecha
s=df["fecha_hora"].astype(str)
mask=s.str.contains("/", na=False)
df["fecha_hora_norm"]=df["fecha_hora"]
df.loc[mask,"fecha_hora_norm"]=pd.to_datetime(df.loc[mask,"fecha_hora"],dayfirst=True,errors="coerce").dt.strftime("%Y-%m-%d %H:%M:%S")
df["fecha_hora_norm"]=pd.to_datetime(df["fecha_hora_norm"],errors="coerce").dt.strftime("%Y-%m-%d %H:%M:%S")
df["fecha_hora"]=df["fecha_hora_norm"]
df.drop(columns=["fecha_hora_norm"],inplace=True)
df.head()
```

```
# Duplicados
dup=df[df.duplicated(keep="first")]
print("duplicados",len(dup))
```

```
# Eliminar duplicados
df_limpio=df.drop_duplicates(keep="first").copy()
print("orig",len(df),"limpio",len(df_limpio))
```

```
# Guardar
df_limpio.to_csv("ventas_limpias.csv",index=False,encoding="utf-8-sig")
df_limpio.to_excel("ventas_limpias.xlsx",index=False)
df_limpio.head()
```

```
# Verificación
import re
print("dup_rest",df_limpio.duplicated().sum())
chk=df_limpio["fecha_hora"].astype(str)
print("mal_fecha",(chk.str.match(r"^\d{4}-\d{2}-\d{2} 00:00:00$",na=False)==False).sum())
```
